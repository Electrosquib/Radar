```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rp = np.load("/Users/levifarinas/Library/Mobile Documents/com~apple~CloudDocs/Projects/SAR Backprojection/SAR Data/SFCW/Tub Center Close/range_profiles.npz")["rp"]
```

```python
def p(rp):
    plt.figure(figsize=(8, 6))
    plt.imshow(
        np.abs(rp),
        aspect="auto",
        origin="lower",
        cmap="jet"
    )

    plt.xlabel("Fast Time (Samples)")
    plt.ylabel("Slow Time (Chirps)")
    plt.title("Raw RP")
    plt.colorbar(label="Magnitude")

    plt.tight_layout()
    plt.show()
```

```python
residual = rp - np.median(rp, axis=0)
```

```python
p(residual)
```
