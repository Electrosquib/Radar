```python
import numpy as np
import matplotlib.pyplot as plt
```



Schroeder Formula

```python
fs = 20e6
buff_size = 400
spacing = 1e6

t = np.arange(buff_size) / fs
tones = np.arange(1e6, fs / 2, spacing)

ibuff, qbuff = np.zeros(buff_size)

for n, tone in enumerate(tones):
    phi = 0
    ibuff += np.sin(2 * np.pi * tone * t + phi)
    qbuff += np.cos(2 * np.pi * tone * t + phi)

plt.plot(t, ibuff)
```
