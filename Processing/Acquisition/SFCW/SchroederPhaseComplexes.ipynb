

Schroeder Formula

## LUT Generation

```python
fs = 20e6
buff_size = 400
spacing = 1e6
t = np.arange(0, buff_size, 1) / fs
tones = np.arange(-fs/2, fs/2, spacing)
buff = np.zeros(buff_size, dtype=np.complex128)
c = 0
for n, f in enumerate(tones):
    phi = -np.pi*n*(n-1)/len(tones) + c
    buff += np.exp(1j * (2 * np.pi * f * t + phi))
ibuff, qbuff = buff.real, buff.imag
plt.plot(t, buff.real)
```
